```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
(X_train_cnn, y_train_cnn), (X_test_cnn, y_test_cnn) = (
    tf.keras.datasets.cifar10.load_data()
)
```

```python
print("Training images:", X_train_cnn.shape)
print("Testing images:", X_test_cnn.shape)
print("Training labels:", y_train_cnn.shape)
print("Testing labels:", y_test_cnn.shape)
```

```python
X_train_cnn = X_train_cnn.astype("float32") / 255.0
X_test_cnn = X_test_cnn.astype("float32") / 255.0 
```

```python
cnn_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(32, 32, 3)),

    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    )
])

cnn_model.summary()
```

```python
cnn_model.add(
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2))
)

cnn_model.summary()
```

```python
cnn_model.add(
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        padding="same",
        activation="relu"
    )
)

cnn_model.summary()
```

```python
cnn_model.add(
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2))
)

cnn_model.summary()
```

```python
cnn_model.add(
    tf.keras.layers.Flatten()
)

cnn_model.summary()
```

```python
cnn_model.add(
    tf.keras.layers.Dense(
        128,
        activation="relu"
    )
)

cnn_model.summary()
```

```python
cnn_model.add(
    tf.keras.layers.Dense(
        10,
        activation="softmax"
    )
)

cnn_model.summary()
```

```python
cnn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("CNN compiled successfully!")
```

```python
# Create a fresh model with the same architecture
cnn_model_early = tf.keras.models.clone_model(cnn_model)

# Configure it for training
cnn_model_early.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn_model_early.summary()
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True,
    verbose=1
)

history_cnn_early = cnn_model_early.fit(
    X_train_cnn,
    y_train_cnn,
    epochs=10,
    batch_size=32,
    validation_split=0.1,
    callbacks=[early_stopping]
)
```
